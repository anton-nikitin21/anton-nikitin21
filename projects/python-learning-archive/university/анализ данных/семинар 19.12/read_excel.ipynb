```python
import pandas as pd
```

```python
pd_SH_2018 = pd.read_excel("ШЧ_2_corr.xlsx", sheet_name="2018", skiprows=7)
print(len(pd_SH_2018))
pd_SH_2018[:1]
```

```python
pd_SH_2019 = pd.read_excel("ШЧ_2_corr.xlsx", sheet_name="2019", skiprows=7)
print(len(pd_SH_2019))
pd_SH_2019[:1]
```

```python
cun = pd.concat([pd_SH_2018,pd_SH_2019], axis=0)
print(len(cun))
cun[:1]
```

```python
print('Количество инцидеентов:', len(cun))
groupby_data = cun.groupby(by =['месяц', 'Станция, перегон'])['Классификация (Группа)'].count()
groupby_data
```

```python
total = 12000
chunksize = 1000
for i, skip in  enumerate(range(0, total, chunksize)):
  print(i, skip)
  if i == 0:
    df = pd.read_excel('ШЧ_2_corr.xlsx', skiprows = skip, nrows = skip + chunksize)
    col_name = df.columns
  else:
    df = pd.read_excel('ШЧ_2_corr.xlsx', skiprows = skip, nrows = chunksize)
    df.columns = col_name
  df.to_excel('ШЧ_2_corr_{}.xlsx'.format(i))
  print('Сохранили часть № {}'.format(i))
```
