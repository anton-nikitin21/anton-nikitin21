```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("Время простоя.xlsx")
df.head()
```

```python
filtered = df[
    (df["Перевозка завершена"] == True) &
    (df["Простой"] > 5)
]
```

```python
total_operators = filtered["Оператор"].nunique()
total_operators
```

```python
operator_pareto = (
    filtered
    .groupby("Оператор")["Простой"]
    .sum()
    .reset_index()
)
```

```python
operator_pareto = operator_pareto.sort_values(
    by="Простой",
    ascending=False
)

operator_pareto["Накопленный %"] = (
    operator_pareto["Простой"].cumsum() /
    operator_pareto["Простой"].sum() * 100
)

operator_pareto
```

```python
top_80 = operator_pareto[operator_pareto["Накопленный %"] <= 80]
top_80
```

```python
share_operators = len(top_80) / total_operators * 100
share_operators
```

```python
mean_downtime_top80 = top_80["Простой"].mean()
mean_downtime_top80
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.bar(
    operator_pareto["Оператор"],
    operator_pareto["Простой"]
)
ax1.set_ylabel("Суммарный простой (дни)")
ax1.set_xlabel("Оператор")
ax1.tick_params(axis="x", rotation=90)

ax2 = ax1.twinx()
ax2.plot(
    operator_pareto["Оператор"],
    operator_pareto["Накопленный %"],
    marker="o"
)
ax2.set_ylabel("Накопленный процент (%)")
ax2.axhline(80, linestyle="--")

plt.title("Диаграмма Парето простоев по операторам")
plt.show()
```
