```python
my_stat = my_table["Станция отправления "].value_counts()
```

```python
my_stat = pd.DataFrame(my_stat)
my_stat
```

```python
my_stat['доля'] = my_stat['count']/my_stat['count'].sum()
my_stat
```

```python
my_stat["Накопленный итог"] = my_stat["доля"].cumsum()*100
my_stat
```

```python
my_stat['cumsum'] = my_stat['доля'].cumsum()
```

```python
my_stat[my_stat['cumsum']<0.8]
```
